```python
df_county = pd.read_excel('Country-Code.xlsx')
df_county
```

```python
final_df = pd.merge(df,df_county,on='Country Code',how='left')
```

```python
final_df.dtypes
```

Observation on the basis of Country

```python
country_name = final_df.Country.value_counts().index
country_value = final_df.Country.value_counts().values
```

```python
country_value
```

```python
# TOP 5 Country uses Zomato
plt.pie(country_value[:5],labels=country_name[:5],autopct='%1.2f')
plt.title("Top 5 County using Zomato")
```

- India,US,UK Brazil and UAE are the top 5 country that uses Zomato
- India is the country with the highest consumption on Zomato, with approximately 93.17%
- The United States is the second-largest consumer of Zomato, accounting for approximately 4.67%.

Observation of Rating 

```python
final_df.columns
```

```python
rating = final_df.groupby(['Aggregate rating','Rating color','Rating text']).size().reset_index().rename(columns={0:'Rating count'})
```

```python
rating
```

## Observation 
1. When rating is between 4.5-4.9 and is in DarkGreen Color ---> Excellent
2. When rating is between 4.0-4.4 and is in Green Color -------> Very Good
3. When rating is between 3.5-3.9 and is in Yellow Color ------_> Good
4. When rating is between 2.5-3.4 and is in Orange Color ------> Average
4. When rating is between 1.8-2.4 and is in Red Color ---------> Poor
4. When there is NO rating or in White Color ------------------> Not Rated

```python
# Creating th Bar plot for the above Observation on the basis of the Color used in the data
plt.figure(figsize=(12,8))
sns.barplot(x='Aggregate rating',y='Rating count',data=rating,hue='Rating color',palette=['lightblue','red','orange','yellow','green','darkgreen'])
plt.title("Overview of Rating by the Customers")
```

### Conclusion
- Not Rated count is very high
- A rating of 3.2 is the most frequently given by customers.
- Maximum number of rating is between 3.0 to 3.9

```python
# Counting the number of rating based on the data 
sns.countplot(x='Rating color',data=rating,palette=['lightblue','red','orange','yellow','green','darkgreen'])
```
